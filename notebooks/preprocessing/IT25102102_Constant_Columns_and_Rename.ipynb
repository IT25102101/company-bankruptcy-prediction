## Constant Column Removal and Column Renaming

*Section owner: Rathnaweera R. P. S. D. (IT25102102)*

Leading spaces are stripped from the column headers. `Net Income Flag` has the same value in every row, so it is dropped as a zero-variance column.

```python
# remove leading/trailing spaces from headers
df.columns = df.columns.str.strip()
df.dtypes
```

```python
# columns with the fewest unique values, to spot constant columns
df.nunique().sort_values().head()
```

```python
# drop the zero-variance column
df = df.drop(['Net Income Flag'], axis=1)
df.shape
```

### Class Distribution

Proportion of non-bankrupt (0) and bankrupt (1) companies.

```python
print("\nClass distribution (0: Not Bankrupt, 1: Bankrupt):")
# proportion of each class
print(df['Bankrupt?'].value_counts(normalize=True))

plt.figure(figsize=(6, 4))
sns.countplot(x=df['Bankrupt?'])
plt.title("Class Distribution (0: Not Bankrupt, 1: Bankrupt)")
plt.show()
```

Only about 3% of companies are bankrupt, so the dataset is highly imbalanced and later outlier removal must preserve the bankrupt rows.
