## Feature Scaling

*Section owner: Raveenath K. A. Y. (IT25102101)*

`MinMaxScaler` rescales all feature columns to the range 0–1, because some columns (e.g. revenue-based figures) have much larger raw values than the ratio columns.

```python
X = df.drop(columns=['Bankrupt?'])   # features
y = df['Bankrupt?']                     # target

scaler = MinMaxScaler()
# scale every feature column
scale_cols = list(X.columns)
X[scale_cols] = scaler.fit_transform(X[scale_cols])
print("Shape:", X.shape)
X.sample(6, random_state=42)
```
